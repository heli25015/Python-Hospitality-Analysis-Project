<h1 style="color:purple"> Data Cleaning </h2>

```python
df_bookings.describe()
```

### 1. Clean Invalid Guests

```python
df_bookings[df_bookings.no_guests<=0]
```

Since these records represent data error and are few in number, we can ignore them

```python
df_bookings = df_bookings[df_bookings.no_guests>0]
```

```python
df_bookings.shape
```

### 2. Outlier removal in revenue generated

```python
df_bookings.revenue_generated.min(), df_bookings.revenue_generated.max()
```

```python
avg, std = df_bookings.revenue_generated.mean(), df_bookings.revenue_generated.std()
avg, std
```

```python
higher_limit = avg + 3*std
lower_limit = avg - 3*std
higher_limit, lower_limit  
```

```python
df_bookings[df_bookings.revenue_generated < 0]
```

```python
df_bookings[df_bookings.revenue_generated > higher_limit]
```

```python
df_bookings = df_bookings[df_bookings.revenue_generated < higher_limit] 
df_bookings.shape
```

```python
df_bookings.revenue_realized.describe()
```

```python
high_limit=df_bookings.revenue_realized.mean() + 3*df_bookings.revenue_realized.std()
high_limit
```

```python
df_bookings[df_bookings.revenue_realized > high_limit]
```

Since, max revenue is 45k and upper limit is 33k there are 1300 outliers and all of these rooms are RT4 (Presidential Suite), we'll perform further analysis

```python
df_bookings[df_bookings.room_category=='RT4'].revenue_realized.describe()
```

Here higher limit comes to be 50583 and in our dataframe above we can see that max value for revenue realized is 45220. Hence we can conclude that there is no outlier and we don't need to do any data cleaning on this particular column

### 3. Handling Null Values

df_bookings.isnull().sum()

Total values in our dataframe is 134576. Out of that 77899 rows has null rating. Since there are many rows with null rating, we should not filter these values. Also we should not replace this rating with a median or mean rating etc

```python
df_agg_bookings.isnull().sum()
```

```python
df_agg_bookings[df_agg_bookings.capacity.isna()]
```

```python
df_agg_bookings.capacity.fillna(df_agg_bookings.capacity.median(), inplace=True)
df_agg_bookings.loc[[8,15]]
```

#### Filtering records that have successful_bookings value greater than capacity

```python
df_agg_bookings[df_agg_bookings.successful_bookings>df_agg_bookings.capacity]
```

```python
df_agg_bookings.shape
```

```python
df_agg_bookings = df_agg_bookings[df_agg_bookings.successful_bookings<=df_agg_bookings.capacity]
df_agg_bookings.shape
```

<h2 style= "color:purple"> Data Transformation </h2>

#### Create an occupancy percentage column

```python
df_agg_bookings.head()
```

```python
df_agg_bookings['occ_pct']=df_agg_bookings['successful_bookings']/df_agg_bookings['capacity']  
df_agg_bookings.head()
```

```python
df_agg_bookings['occ_pct']=df_agg_bookings['occ_pct'].apply(lambda x: round(x*100,2))
df_agg_bookings.head()
```
